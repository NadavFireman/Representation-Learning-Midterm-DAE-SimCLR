<div dir="ltr" align="left" style="font-family: Calibri, sans-serif; font-size: 12pt; line-height: 1.5;">
<p>Step 5: one run of train.py</p>
<p>Set MODE to one of these and run the cell:</p>
<ul>
<li>pretrain: trains the denoising autoencoder on Tiny-ImageNet without labels, and saves its encoder.</li>
<li>pretrain_supervise_encoder: the same encoder trained from scratch with labels (the supervised baseline).</li>
<li>linear_probe: the saved encoder stays frozen, and only a new linear head is trained.</li>
<li>finetune: the saved encoder and a new head are trained together, with a smaller learning rate.</li>
<li>knn: no training. k-NN accuracy (k = 20, cosine) of the saved encoder.</li>
<li>analysis: no training. Per-class k-NN accuracy, t-SNE, nearest-neighbor retrieval and the robustness test of the saved encoder (analysis.py).</li>
</ul>
<p>Other flags can be added to the command, for example:</p>
<ul>
<li>--noise_type gaussian, mask or salt_pepper, --noise_level and --latent_dim: the DAE hyperparameters.</li>
<li>--label_fraction 1.0, 0.1 or 0.01: the class-balanced share of labels for linear_probe, finetune and the supervised baseline.</li>
<li>--encoder dae, supervised or random: which encoder linear_probe, finetune, knn and analysis load.</li>
<li>--dataset cifar100: evaluate on the transfer dataset.</li>
<li>--search: one trial of the hyperparameter search (proxy setting, no test set, one row in sweep_results.csv).</li>
<li>--seed: the training seed, for repeat runs.</li>
</ul>
<p>The test results are printed at the end of the run, and all the curves are in TensorBoard.</p>

<div dir="ltr" align="left" style="font-family: Calibri, sans-serif; font-size: 12pt; line-height: 1.5;">
<p>Step 6: all the runs of Part A</p>
<ul>
<li>Runs train.py once per line of RUNS, in order: the encoders first, then the evaluation protocol for each encoder, then the analysis.</li>
<li>Encoders: the DAE, the plain autoencoder (the same DAE with --noise_level 0) and the supervised encoder at 100%, 10% and 1% of the labels. The random-init encoder needs no training.</li>
<li>Protocol, for the DAE, the AE and the random encoder, on Tiny-ImageNet and on CIFAR-100 (transfer): k-NN, linear probe, and fine-tuning with 1%, 10% and 100% of the labels.</li>
<li>The supervised encoder trained with all the labels also gets k-NN and the linear probe on both datasets. Its fine-tuning numbers on Tiny-ImageNet are the test results of its own pretrain run.</li>
<li>analysis, for every encoder on Tiny-ImageNet: per-class accuracy, t-SNE, retrieval and the robustness test.</li>
<li>After a disconnect, run the cell again. A run that already finished wrote results.json in its folder, so it only prints its results and is skipped; an unfinished run continues from its last.ckpt.</li>
<li>The classifier runs (the supervised encoder, linear probe and fine-tuning) use config.PROTOCOL "acc": they stop on and keep the best val_accuracy, with patience 5 and label smoothing 0.1, and their folders end in _acc. The first runs, chosen by val_loss, keep their own folders next to them, untouched. The DAE and the AE are chosen by val k-NN either way, so they are not trained again.</li>
</ul>

<div dir="rtl" align="right" style="font-family: Calibri, sans-serif; font-size: 12pt; line-height: 1.5;">
<p>שלב 6: כל הריצות של חלק A</p>
<ul>
<li>מריץ את train.py פעם אחת לכל שורה ב-RUNS, לפי הסדר: קודם ה-encoders, אחר כך פרוטוקול ההערכה לכל encoder, ובסוף ה-analysis.</li>
<li>‏Encoders: ה-DAE, ה-autoencoder הרגיל (אותו DAE עם ‎--noise_level 0) וה-encoder המפוקח עם 100%, ‏10% ו-1% מהתוויות. ה-encoder האקראי לא צריך אימון.</li>
<li>הפרוטוקול, ל-DAE, ל-AE ול-encoder האקראי, על Tiny-ImageNet ועל CIFAR-100 (‏transfer): ‏k-NN, ‏linear probe ו-fine-tuning עם 1%, ‏10% ו-100% מהתוויות.</li>
<li>ה-encoder המפוקח שאומן עם כל התוויות מקבל גם k-NN ו-linear probe על שני הדאטה-סטים. תוצאות ה-fine-tuning שלו על Tiny-ImageNet הן תוצאות ה-test של ריצת ה-pretrain שלו.</li>
<li>‏analysis, לכל encoder על Tiny-ImageNet: דיוק לכל מחלקה, ‏t-SNE, שליפת שכנים ומבחן העמידות.</li>
<li>אחרי ניתוק מריצים את התא שוב. ריצה שכבר הסתיימה כתבה results.json בתיקייה שלה, ולכן היא רק מדפיסה את התוצאות שלה ומדלגים עליה; ריצה שלא הסתיימה ממשיכה מה-last.ckpt שלה.</li>
<li>ריצות המסווג (ה-encoder המפוקח, ‏linear probe ו-fine-tuning) משתמשות ב-config.PROTOCOL ‏"acc": הן נעצרות לפי ה-val_accuracy הכי טוב ושומרות אותו, עם patience 5 ו-label smoothing 0.1, והתיקיות שלהן מסתיימות ב-_acc. הריצות הראשונות, שנבחרו לפי val_loss, נשארות בתיקיות שלהן לידן, בלי שינוי. ה-DAE וה-AE נבחרים לפי val k-NN בשני המקרים, ולכן הם לא מתאמנים שוב.</li>
</ul>
</div>

```python
# Every run of Part A. Each line is one call of train.py.
ENCODERS = {  # the flags that pick each encoder
    "dae": "--encoder dae",
    "ae": "--encoder dae --noise_level 0",  # the plain autoencoder: the same DAE, no corruption
    "random": "--encoder random",
}
FRACTIONS = (0.01, 0.1, 1.0)
DATASETS = ("tinyimagenet", "cifar100")

RUNS = [
    "--mode pretrain",  # the DAE
    "--mode pretrain --noise_level 0",  # the AE baseline
    *[f"--mode pretrain_supervise_encoder --label_fraction {f}" for f in FRACTIONS],  # the supervised baseline
]
for flags in ENCODERS.values():
    for dataset in DATASETS:
        RUNS += [
            f"--mode knn {flags} --dataset {dataset}",
            f"--mode linear_probe {flags} --dataset {dataset}",
            *[f"--mode finetune {flags} --dataset {dataset} --label_fraction {f}" for f in FRACTIONS],
        ]
for dataset in DATASETS:  # the supervised encoder (all labels) under the frozen protocol
    RUNS += [
        f"--mode knn --encoder supervised --dataset {dataset}",
        f"--mode linear_probe --encoder supervised --dataset {dataset}",
    ]
RUNS += [f"--mode analysis {flags}" for flags in (*ENCODERS.values(), "--encoder supervised")]

for i, flags in enumerate(RUNS, 1):
    print(f"\n=== run {i}/{len(RUNS)}: {flags}")
    !python train.py {flags} --output_dir "{OUTPUT_DIR}"
```
