<p>Step 7: stage 2, the full protocol with ResNet-18</p>
<ul>
<li>The same runs as step 6 of colab_train.ipynb, with the ResNet-18 backbone, so the ResNet tables line up with the main ones.</li>
<li>Encoders: the AE (the chosen DAE with --noise_level 0), and the supervised encoder at 10% and 1% of the labels (100% ran in stage 1).</li>
<li>Protocol, for the DAE, the AE and the random encoder, on Tiny-ImageNet and on CIFAR-100: k-NN, linear probe, and fine-tuning with 1%, 10% and 100% of the labels.</li>

<li>Runs that stage 1 already finished are skipped (their results.json). This stage is long; it can be stopped and started again at any point.</li>

<div dir="rtl" align="right" style="font-family: Calibri, sans-serif; font-size: 12pt; line-height: 1.5;">
<p>שלב 7: שלב 2, הפרוטוקול המלא עם ResNet-18</p>
<ul>
<li>אותן ריצות כמו שלב 6 של colab_train.ipynb, עם ה-backbone של ResNet-18, כך שהטבלאות של ResNet מקבילות לטבלאות הראשיות.</li>
<li>‏Encoders: ה-AE (ה-DAE שנבחר עם ‎--noise_level 0), וה-encoder המפוקח עם 10% ו-1% מהתוויות (100% רץ בשלב 1).</li>
<li>הפרוטוקול, ל-DAE, ל-AE ול-encoder האקראי, על Tiny-ImageNet ועל CIFAR-100: ‏k-NN, ‏linear probe ו-fine-tuning עם 1%, ‏10% ו-100% מהתוויות.</li>
<li>ה-encoder המפוקח (כל התוויות) מקבל k-NN ו-linear probe על שני הדאטה-סטים. תוצאות ה-fine-tuning שלו על Tiny-ImageNet הן תוצאות ה-test של ריצות ה-pretrain שלו.</li>
<li>‏analysis לארבעת ה-encoders: דיוק לכל מחלקה, ‏t-SNE, שליפת שכנים ומבחן העמידות.</li>
<li>ריצות ששלב 1 כבר סיים מדולגות (לפי ה-results.json שלהן). השלב הזה ארוך; אפשר לעצור ולהתחיל אותו מחדש בכל נקודה.</li>
</ul>
</div>

```python
assert DECODER in ("light", "resnet18"), "Run step 6 first."
ENCODERS = {  # the flags that pick each encoder
    "dae": f"--encoder dae {R}{D}",
    "ae": f"--encoder dae --noise_level 0 {R}{D}",  # the plain autoencoder: the same DAE, no corruption
    "random": f"--encoder random {R}",
}
SUPERVISED = f"--encoder supervised {R}"
FRACTIONS = (0.01, 0.1, 1.0)
DATASETS = ("tinyimagenet", "cifar100")

RUNS = [
    f"--mode pretrain --noise_level 0 {R}{D}",  # the AE baseline
    *[f"--mode pretrain_supervise_encoder --label_fraction {f} {R}" for f in FRACTIONS],  # the supervised baseline
]
for flags in ENCODERS.values():
    for dataset in DATASETS:
        RUNS += [
            f"--mode knn {flags} --dataset {dataset}",
            f"--mode linear_probe {flags} --dataset {dataset}",
            *[f"--mode finetune {flags} --dataset {dataset} --label_fraction {f}" for f in FRACTIONS],
        ]
for dataset in DATASETS:  # the supervised encoder (all labels) under the frozen protocol
    RUNS += [
        f"--mode knn {SUPERVISED} --dataset {dataset}",
        f"--mode linear_probe {SUPERVISED} --dataset {dataset}",
    ]
RUNS += [f"--mode analysis {flags}" for flags in (*ENCODERS.values(), SUPERVISED)]

for i, flags in enumerate(RUNS, 1):
    print(f"\n=== stage 2, run {i}/{len(RUNS)}: {flags}")
    !python train.py {flags} --output_dir "{OUTPUT_DIR}"
```
